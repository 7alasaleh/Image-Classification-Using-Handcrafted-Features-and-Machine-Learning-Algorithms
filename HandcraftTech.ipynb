```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Preprocessing function
def preprocess_image(image_path, size=(224, 224)):

    # Read image as RGB
    image = cv2.imread(image_path)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Resize RGB image
    resized_rgb = cv2.resize(
        image,
        size,
        interpolation=cv2.INTER_AREA
    )

    # Convert to Grayscale
    grayscale = cv2.cvtColor(
        resized_rgb,
        cv2.COLOR_RGB2GRAY
    )

    return image, resized_rgb, grayscale


# Select one sample from each class
fig, axes = plt.subplots(6, 3, figsize=(12, 22))

for i, class_name in enumerate(classes):

    sample = train_df[
        train_df["class_name"] == class_name
    ].sample(n=1, random_state=42).iloc[0]

    original, resized, gray = preprocess_image(sample["path"])

    axes[i, 0].imshow(original)
    axes[i, 0].set_title("Original RGB (256x256)")

    axes[i, 1].imshow(resized)
    axes[i, 1].set_title("Resized RGB (224x224)")

    axes[i, 2].imshow(gray, cmap="gray")
    axes[i, 2].set_title("Grayscale (224x224)")

    for ax in axes[i]:
        ax.axis("off")

    print(
        f"{class_name}: "
        f"Original={original.shape}, "
        f"Resized={resized.shape}, "
        f"Grayscale={gray.shape}"
    )

plt.tight_layout()
plt.show()
```

```python
from skimage.feature import local_binary_pattern
from tqdm.notebook import tqdm

P = 8
R = 1
METHOD = "uniform"
N_BINS = P + 2


def extract_lbp_features(image_path):
    _, _, gray = preprocess_image(image_path)

    lbp_image = local_binary_pattern(
        gray,
        P=P,
        R=R,
        method=METHOD
    )

    counts, _ = np.histogram(
        lbp_image.ravel(),
        bins=np.arange(N_BINS + 1)
    )

    lbp_hist = counts.astype(np.float64)
    lbp_hist /= lbp_hist.sum()

    return lbp_hist


def extract_features_from_dataframe(df, description):
    features = []

    for path in tqdm(df["path"], desc=description):
        feature_vector = extract_lbp_features(path)
        features.append(feature_vector)

    return np.array(features)


X_train_lbp = extract_features_from_dataframe(
    train_df, "Extracting LBP - Train"
)

X_val_lbp = extract_features_from_dataframe(
    val_df, "Extracting LBP - Validation"
)

X_test_lbp = extract_features_from_dataframe(
    test_df, "Extracting LBP - Test"
)


y_train = train_df["class_name"].to_numpy()
y_val = val_df["class_name"].to_numpy()
y_test = test_df["class_name"].to_numpy()
```

```python
from skimage.feature import graycomatrix, graycoprops
from tqdm.notebook import tqdm

GLCM_DISTANCES = [1, 2]
GLCM_ANGLES = [0, np.pi/4, np.pi/2, 3*np.pi/4]
GLCM_LEVELS = 32

GLCM_PROPERTIES = [
    "contrast",
    "dissimilarity",
    "homogeneity",
    "energy",
    "correlation"
]


def extract_glcm_features(image_path):

    _, _, gray = preprocess_image(image_path)

    # Convert grayscale image from 256 levels to 32 levels
    gray_quantized = (gray // 8).astype(np.uint8)

    glcm = graycomatrix(
        gray_quantized,
        distances=GLCM_DISTANCES,
        angles=GLCM_ANGLES,
        levels=GLCM_LEVELS,
        symmetric=True,
        normed=True
    )

    features = []

    for prop in GLCM_PROPERTIES:
        values = graycoprops(glcm, prop)
        features.extend(values.flatten())

    return np.array(features, dtype=np.float32)


def extract_glcm_from_dataframe(df, description):

    features = np.empty(
        (len(df), 40),
        dtype=np.float32
    )

    for i, path in enumerate(tqdm(df["path"], desc=description)):
        features[i] = extract_glcm_features(path)

    return features


X_train_glcm = extract_glcm_from_dataframe(
    train_df, "Extracting GLCM - Train"
)

X_val_glcm = extract_glcm_from_dataframe(
    val_df, "Extracting GLCM - Validation"
)

X_test_glcm = extract_glcm_from_dataframe(
    test_df, "Extracting GLCM - Test"
)
```

```python
def create_combined_features(
    X_lbp,
    X_hog,
    X_glcm,
    split_name
):
    total_features = (
        X_lbp.shape[1]
        + X_hog.shape[1]
        + X_glcm.shape[1]
    )

    output_path = os.path.join(
        FEATURE_DIR,
        f"X_{split_name}_combined.npy"
    )

    combined = np.lib.format.open_memmap(
        output_path,
        mode="w+",
        dtype=np.float32,
        shape=(len(X_lbp), total_features)
    )

    lbp_end = X_lbp.shape[1]
    hog_end = lbp_end + X_hog.shape[1]

    combined[:, :lbp_end] = X_lbp

    combined[:, lbp_end:hog_end] = X_hog

    combined[:, hog_end:] = X_glcm

    combined.flush()

    return combined


X_train_combined = create_combined_features(
    X_train_lbp,
    X_train_hog,
    X_train_glcm,
    "train"
)

X_val_combined = create_combined_features(
    X_val_lbp,
    X_val_hog,
    X_val_glcm,
    "validation"
)

X_test_combined = create_combined_features(
    X_test_lbp,
    X_test_hog,
    X_test_glcm,
    "test"
)

print("Combined feature extraction completed.")
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    precision_score,
    recall_score
)

import time

# Store trained models and validation results
trained_models = {}
validation_results = []


# Define LBP classifiers
lbp_classifiers = {
    "SVM": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", LinearSVC(
            C=1.0,
            max_iter=10000,
            random_state=42
        ))
    ]),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        max_features="sqrt",
        random_state=42,
        n_jobs=-1
    ),

    "KNN": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", KNeighborsClassifier(
            n_neighbors=5,
            weights="distance",
            n_jobs=-1
        ))
    ])
}


# Train and evaluate LBP classifiers
for model_name, model in lbp_classifiers.items():

    print(f"\nTraining LBP + {model_name}...")

    start_time = time.time()

    model.fit(X_train_lbp, y_train)

    y_val_pred = model.predict(X_val_lbp)

    elapsed_time = time.time() - start_time

    accuracy = accuracy_score(y_val, y_val_pred)
    macro_f1 = f1_score(
        y_val,
        y_val_pred,
        average="macro"
    )

    macro_precision = precision_score(
        y_val,
        y_val_pred,
        average="macro",
        zero_division=0
    )

    macro_recall = recall_score(
        y_val,
        y_val_pred,
        average="macro",
        zero_division=0
    )

    trained_models[("LBP", model_name)] = model

    validation_results.append({
        "Feature Set": "LBP",
        "Classifier": model_name,
        "Accuracy": accuracy,
        "Macro Precision": macro_precision,
        "Macro Recall": macro_recall,
        "Macro F1": macro_f1,
        "Time (seconds)": elapsed_time
    })

    print(f"Accuracy: {accuracy:.4f}")
    print(f"Macro F1: {macro_f1:.4f}")
    print(f"Time: {elapsed_time:.2f} seconds")
```

```python
import time
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# Test feature sets
test_features = {
    "LBP": X_test_lbp,
    "HOG": X_test_hog,
    "GLCM": X_test_glcm,
    "Combined": X_test_combined
}

test_results = []
test_predictions = {}

print("Starting Final Test Evaluation...\n")

for feature_name, X_test in test_features.items():

    for classifier_name in ["SVM", "Random Forest", "KNN"]:

        model_key = (feature_name, classifier_name)
        model = trained_models[model_key]

        print(f"Evaluating {feature_name} + {classifier_name}...")

        # Prediction only - no training
        start_time = time.perf_counter()
        y_pred = model.predict(X_test)
        prediction_time = time.perf_counter() - start_time

        # Save predictions
        test_predictions[model_key] = y_pred

        # Evaluation metrics
        accuracy = accuracy_score(y_test, y_pred)

        macro_precision = precision_score(
            y_test, y_pred,
            average="macro",
            zero_division=0
        )

        macro_recall = recall_score(
            y_test, y_pred,
            average="macro",
            zero_division=0
        )

        macro_f1 = f1_score(
            y_test, y_pred,
            average="macro",
            zero_division=0
        )

        weighted_f1 = f1_score(
            y_test, y_pred,
            average="weighted",
            zero_division=0
        )

        test_results.append({
            "Feature Set": feature_name,
            "Classifier": classifier_name,
            "Accuracy": accuracy,
            "Macro Precision": macro_precision,
            "Macro Recall": macro_recall,
            "Macro F1": macro_f1,
            "Weighted F1": weighted_f1,
            "Prediction Time": prediction_time
        })

        print(
            f"Accuracy: {accuracy:.4f} | "
            f"Macro F1: {macro_f1:.4f} | "
            f"Prediction Time: {prediction_time:.3f}s\n"
        )

# Create final results DataFrame
test_results_df = pd.DataFrame(test_results)

test_results_df = test_results_df.sort_values(
    by="Macro F1",
    ascending=False
).reset_index(drop=True)

print("\nFINAL TEST RESULTS")
display(
    test_results_df.style.format({
        "Accuracy": "{:.4%}",
        "Macro Precision": "{:.4%}",
        "Macro Recall": "{:.4%}",
        "Macro F1": "{:.4%}",
        "Weighted F1": "{:.4%}",
        "Prediction Time": "{:.3f} s"
    })
)

# Save results and predictions
test_results_df.to_csv(
    "/kaggle/working/handcrafted_test_results.csv",
    index=False
)

print("\nTest evaluation completed.")
print("Results saved to handcrafted_test_results.csv")
print("Number of saved prediction sets:", len(test_predictions))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    confusion_matrix,
    classification_report
)

# Models selected using Validation results
selected_models = {
    "LBP + RF": ("LBP", "Random Forest"),
    "HOG + RF": ("HOG", "Random Forest"),
    "GLCM + RF": ("GLCM", "Random Forest"),
    "Combined + RF": ("Combined", "Random Forest")
}

class_names = sorted(np.unique(y_test))

# ==========================================
# 1. Confusion Matrices
# ==========================================

fig, axes = plt.subplots(2, 2, figsize=(18, 15))

for ax, (model_name, model_key) in zip(
    axes.flat, selected_models.items()
):

    y_pred = test_predictions[model_key]

    cm = confusion_matrix(
        y_test,
        y_pred,
        labels=class_names
    )

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=class_names,
        yticklabels=class_names,
        ax=ax
    )

    ax.set_title(model_name, fontsize=14, fontweight="bold")
    ax.set_xlabel("Predicted Label")
    ax.set_ylabel("True Label")
    ax.tick_params(axis="x", rotation=45)
    ax.tick_params(axis="y", rotation=0)

plt.suptitle(
    "Confusion Matrices - Final Test Evaluation",
    fontsize=18,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    "/kaggle/working/final_test_confusion_matrices.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ==========================================
# 2. Classification Reports
# ==========================================

all_reports = []

for model_name, model_key in selected_models.items():

    y_pred = test_predictions[model_key]

    report = classification_report(
        y_test,
        y_pred,
        labels=class_names,
        target_names=class_names,
        output_dict=True,
        zero_division=0
    )

    report_df = pd.DataFrame(report).T
    report_df.index.name = "Class"

    print(f"\n{'='*70}")
    print(f"Classification Report: {model_name}")
    print("="*70)

    display(
        report_df.style.format({
            "precision": "{:.4f}",
            "recall": "{:.4f}",
            "f1-score": "{:.4f}",
            "support": "{:.0f}"
        })
    )

    for class_name in class_names:
        all_reports.append({
            "Model": model_name,
            "Class": class_name,
            "Precision": report[class_name]["precision"],
            "Recall": report[class_name]["recall"],
            "F1-Score": report[class_name]["f1-score"],
            "Support": report[class_name]["support"]
        })


# ==========================================
# 3. One-vs-Rest Metrics
# ==========================================

ovr_results = []

for model_name, model_key in selected_models.items():

    y_pred = test_predictions[model_key]

    cm = confusion_matrix(
        y_test,
        y_pred,
        labels=class_names
    )

    for i, class_name in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = cm.sum() - TP - FN - FP

        sensitivity = TP / (TP + FN) if TP + FN else 0
        specificity = TN / (TN + FP) if TN + FP else 0
        fpr = FP / (FP + TN) if FP + TN else 0
        fnr = FN / (FN + TP) if FN + TP else 0
        fdr = FP / (FP + TP) if FP + TP else 0

        ovr_results.append({
            "Model": model_name,
            "Class": class_name,
            "Sensitivity": sensitivity,
            "Specificity": specificity,
            "FPR": fpr,
            "FNR": fnr,
            "FDR": fdr
        })


# Convert results to DataFrames
classification_reports_df = pd.DataFrame(all_reports)
ovr_metrics_df = pd.DataFrame(ovr_results)

# Save reports
classification_reports_df.to_csv(
    "/kaggle/working/classification_reports.csv",
    index=False
)

ovr_metrics_df.to_csv(
    "/kaggle/working/one_vs_rest_metrics.csv",
    index=False
)

print("\nAll reports and confusion matrices saved successfully.")
```
